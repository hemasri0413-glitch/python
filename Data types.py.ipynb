```python
#keys, value, item, get,update, pop method
student = {
    "name" : "Hemasri",
    "age"  : 18,
    "course": "python"
}
print(student.keys())
#keys() retuns all keys values

print(student.values())
#values() returns all the values in the dictionary

print(student.items())
#items() returns all key-value pairs

print(student.get("name"))
#get() returns the value of 

print(student.update({"age":18}))
#update() updates the values of the specified key

print(student)

student.pop("age")
#pop() removes the specified key

print(student)
```
