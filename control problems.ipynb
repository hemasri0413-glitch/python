```python
#Reverse the numbers 
number = int(input("Enter a number:"))

reverse = 0

while number > 0:
    digit = number % 10
    number = number // 10
    reverse = reverse * 10 + digit

print("Reverse:", reverse)
```

```python
#Check if the number is a palindrome
number = int(input("Enter a number:"))

original = number
reverse = 0

while number > 0:
    digit = number % 10
    reverse = reverse * 10 + digit
    number = number // 10

if original == reverse:
        print("Palindrome")
else:
      print("Not palindrome")
```

```python
#Check if a number is prime
number = int(input("Enter a number:"))

count = 0

for i in range(1, number + 1):
    if number % i == 0:
        count = count + 1

if count == 2:
    print("Prime number")
else:
    print("Not a prime number")
```
